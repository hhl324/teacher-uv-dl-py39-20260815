```python
import torch
```

```python
t1 = torch.randint(1, 10, (2, 3, 4)).float()
print(t1)
```

```python
print(t1.sum())
```

```python
print(t1.sum(dim=0))
```

```python
print(t1.sum(dim=1))
```

```python
print(t1.sum(dim=2))
```

```python
# mean
print(t1.mean(dim=0))
```

```python
print(t1.max())
```

```python
print(t1.max(dim=0))
```

```python
print(t1.var())
print(t1.var(dim=0))
```

```python
print(t1.std())
```

```python
print(t1.unique())
```

```python
print(t1.sort(dim=2))
print(t1.sort())  # 不加dim默认是按照矩阵的行进行排序
```
