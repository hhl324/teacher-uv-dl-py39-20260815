```python
import torch
import numpy as np
```

## 1. 基于内容创建tensor

```python
# 传入标量
t1 = torch.tensor(10)
print(t1)
print(t1.shape)
```

```python
# 传入列表 向量
t1 = torch.tensor([10, 20, 30])
print(t1)
print(t1.shape)
```

```python
# 传入一个二维的列表(矩阵)
t1 = torch.tensor([[10, 20, 30], [40, 50, 60]])
print(t1)
print(t1.shape)
```

```python
# 传入一个3维的列表(张量)
t1 = torch.tensor([ [[10, 20, 30], [40, 50, 60]],   [[10, 20, 30], [40, 50, 60]]])
print(t1)
print(t1.shape)
```

```python
t1 = torch.tensor(np.array([[10.0, 20, 30], [40, 50, 60]]))
print(t1)
print(t1.shape)
```

```python
# 查看数据类型
print(t1.dtype)
```

# 4. 指定区间创建张量

```python
t1 = torch.arange(10)  # start=0 end=10 step=1  [start, end)
print(t1)

t1 = torch.arange(start=10, end=100, step=5)  # start=0 end=10 step=1  [start, end)
print(t1)
```

```python
t1 = torch.linspace(1, 10, 3)  # 步长会自动计算: (end-start)/(steps-1)   [start, end]
print(t1)
```

```python
t1 = torch.logspace(1, 3, 3, base=torch.e)  # 10^1  10^2
print(t1)
```

## 5. 根据填充创建张量

```python
t1 = torch.ones(2,3)
print(t1)
```

```python
t2 = torch.ones_like(t1)
print(t2)
```

```python
t1 = torch.zeros(2,3)
print(t1)
t2 = torch.zeros_like(t1)
print(t1)
```

```python
t1 = torch.empty(2,3)
print(t1)
```

```python
t1 = torch.full( (2,3), 10)
print(t1)
```

```python
t1 = torch.eye(5)
print(t1)
```

# 6. 随机创建张量

```python
t1 = torch.rand(2, 3)  # 均匀分布: [0,1]
print(t1)
```

```python
t1 = torch.randint(low=10, high=100, size=(2,3))  # 均匀分布
print(t1)
```

```python
# 标准正态分布
t1 = torch.randn(2, 3)
print(t1)
```

```python
t1 = torch.normal(mean=5, std=3, size=(2,3))
print(t1)
```

```python
# 洗牌(打乱)
t1 = torch.randperm(10)  # [0, 10)
print(t1)
```

```python
# 查看随机数种子
print(torch.random.initial_seed())
```

```python
# 设置随机数种子
torch.random.manual_seed(42)
```
