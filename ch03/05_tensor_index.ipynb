```python
import torch
t = torch.randint(0, 10, (2, 5, 4))
print(t)
```

## 简单索引

```python
print(t[0][0][0])
print(t[0,0,0])
```

```python
print(t[0])  # 取第一个矩阵
```

```python
print(t[0,0])
```

## 范围索引

```python
print(t[1:])
```

```python
print(t[1,1:, 2:])
```

```python
print(t[-2,1:, 2:])
```

## 列表索引

```python
list1 = [0,1,1]
list2 = [1,3,2]
print(t[list1, list2])
```

```python
list1 = [[0], [1]]
list2 = [1,3,2]
print(t[list1, list2])
```

## 布尔索引

```python
#  取出所有大于5的元素
mask = t > 5
mask
```

```python
print(t[mask])
```

```python
# 取出每行的首元素大于5的那些行
mask = t[:, :, 0] > 5
print(mask)

print(t[mask])
```

```python
# 取出每列的首元素大于5的那些列
mask = t[:, 0, :] > 5
print(mask)
```

```python
print(t.mT[mask].T)
```
