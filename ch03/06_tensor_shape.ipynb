```python
import torch
t = torch.randint(0, 10, (2, 5, 4))
print(t)
```

## 维度的变换

```python
print(t.T)  # (0,2,3) => (3,2,0)
```

```python
print(t.mT)  # 只转置最后2个维度
```

```python
# 任意交换两个维度
print(t.transpose(0, 1))
```

```python
# 重新排列所有维度
print(t.permute(0, 2,1))
print(t.permute(2, 0, 1).shape)
```

## 形状调整

```python
t1 = t.reshape(1, -1)
print(t1)
print(t1.shape)
```

```python
t1 = t.reshape(1,1,1,2,-1)
print(t1)
print(t1.shape)
```

```python
# view
print(t1.view(1,-1))

t2 = t1.mT
print(t1.is_contiguous())
print(t2.is_contiguous())
# contiguous 强制内存连续. 一般推荐使用reshape
print(t2.contiguous().view(1,-1))
```

# 增删维度

```python
t1 = torch.randint(1, 10, (2, 4))
print(t1)
print(t1.reshape(2,1, 4))
```

```python
print(t1.unsqueeze(dim=0)) # 增加维度
```

```python
t2 = t1.unsqueeze(dim=0).unsqueeze(dim=0).unsqueeze(dim=-1).unsqueeze(dim=-1)
print(t2)
print(t2)
print(t2.shape)
```

```python
print(t2.squeeze())
```

## tensor的拼接

```python
t1 = torch.randint(1, 10, (2, 1, 3))
t2 = torch.randint(1, 10, (2, 3, 3))
print(t1)
print(t2)
```

```python
print(torch.cat((t1, t2), dim=1))
```

```python
# stack
t1 = torch.randint(1, 10, (2, 2, 5))
t2 = torch.randint(1, 10, (2, 2, 5))
print(t1)
print(t2)
```

```python
# 等价于: 先unsequze, 然后再cat
t3 = torch.stack((t1, t2), dim=-1)
print(t3)
print(t3.shape)
```
