```python
import torch
```

## 1. 元素类型的转换

```python
t1 = torch.tensor([1,2,3])
print(t1.dtype)
```

```python
# 使用type方法
t2 = t1.type(torch.float32)
print(t2.dtype)
```

```python
# to 方法
t2 = t1.to(torch.float32)
print(t2.dtype)
```

```python
# 直接调用类型方法
t2 = t1.float()
print(t2.dtype)
t2 = t1.long()
print(t2.dtype)
```

## tensor =>  ndarray

```python
import numpy as np
```

```python
t1 = torch.tensor([1,2,3])
```

```python
n1 = t1.numpy()
print(type(n1))
print(n1)
```

```python
# 内存的问题: 内存共享
t1[0] = 10
print(t1)
print(n1)
```

```python
n2 = t1.numpy().copy()
print(n2)
t1[0] = 100
print(n2)
```

## ndarray => numpy

```python
n1 = np.array([1,2,3])
```

```python
t1 = torch.from_numpy(n1)
print(t1)
```

```python
n1[0] = 100
print(n1)
print(t1)
```

```python
t2 = torch.from_numpy(n1).clone()
print(t2)
n1[0] = 200
print(n1)
print(t2)
```
