## 2. 元素级运算(张量和张量)

```python
t1 = torch.tensor([[1, 2, 3], [10, 20, 30]])
t2 = torch.tensor([[100, 200, 300], [5, 6, 7]])
t3 = torch.tensor([[100], [5]])
```

```python
# 张量 + 张量  形状一样, 或者满足广播条件
print(t1 + t2)
print(t1 + t3)
```

```python
# 张量 * 张量  对位相差  哈达玛积
print(t1 * t2)
```

## 3. 矩阵相乘

```python
t1 = torch.tensor([[1, 2, 3], [10, 20, 30]])
t2 = torch.tensor([[100, 200, 300], [5, 6, 7], [7, 8, 9]])
```

```python
print(t1 @ t2)
```

```python
print(t1.mm(t2))  # 只能用于矩阵, 维数超过2就不支持
```

```python
print(t1.matmul(t2))
```

```python
t1 = torch.tensor([[[1, 2, 3], [4, 5, 6], [6, 5, 4]], [[3, 2, 1], [3, 2, 1], [3, 2, 1]]])
t2 = torch.tensor([[[1, 2], [3, 4], [5, 6]], [[6, 5], [4, 3], [2, 1]]])
print(t1.shape)
print(t2.shape)
```

```python
print(t1 @ t2)
```
